```python
app_py_content = f"""
import gradio as gr
import torch
import torch.nn as nn
from torchvision import models, transforms
from huggingface_hub import hf_hub_download
from PIL import Image
import numpy as np

# --- 1. CONFIGURATION ---
MODEL_REPO = "{MODEL_REPO_ID}"
MODEL_FILE = "{BEST_MODEL_PATH}"

# CIFAR-100 class names
CIFAR100_CLASSES = {CIFAR100_CLASSES}

# CIFAR-100 Normalization stats
CIFAR100_MEAN = (0.5071, 0.4867, 0.4408)
CIFAR100_STD  = (0.2675, 0.2565, 0.2761)

# --- 2. MODEL DEFINITION ---
def resnet18_cifar(num_classes=100):
    model = models.resnet18(pretrained=False)
    model.conv1 = nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1, bias=False)
    model.maxpool = nn.Identity()
    model.fc = nn.Linear(model.fc.in_features, num_classes)
    return model

# --- 3. MODEL LOADING ---
def load_model():
    # Download the checkpoint file
    model_path = hf_hub_download(repo_id=MODEL_REPO, filename=MODEL_FILE)

    # Initialize model and load weights
    model = resnet18_cifar(num_classes=100)
    checkpoint = torch.load(model_path, map_location=torch.device('cpu'))
    model.load_state_dict(checkpoint['model'])
    model.eval()

    # Define the image preprocessor (used for inference)
    preprocessor = transforms.Compose([
        transforms.Resize(32),
        transforms.CenterCrop(32),
        transforms.ToTensor(),
        transforms.Normalize(CIFAR100_MEAN, CIFAR100_STD)
    ])

    return model, preprocessor

model, preprocessor = load_model()

# --- 4. PREDICTION FUNCTION ---
def predict_image(image: np.ndarray):
    # Convert numpy array (from Gradio) to PIL Image
    pil_image = Image.fromarray(image.astype('uint8'))

    # Preprocess the image
    input_tensor = preprocessor(pil_image)
    input_batch = input_tensor.unsqueeze(0)  # create a mini-batch as expected by the model

    with torch.no_grad():
        output = model(input_batch)

    probabilities = torch.nn.functional.softmax(output[0], dim=0)

    # Get top 5 predictions
    top5_prob, top5_catid = torch.topk(probabilities, 5)

    # Format for Gradio output
    results = {{}}
    for i in range(top5_prob.size(0)):
        class_name = CIFAR100_CLASSES[top5_catid[i].item()]
        results[class_name] = top5_prob[i].item()

    return results

# --- 5. GRADIO INTERFACE ---
title = "ResNet18 (Scratch) on CIFAR-100"
description = f"Image classification demo using a custom-trained ResNet18 model (Achieved {BEST_VAL:.2f}% Top-1 Accuracy)."
article = "Model weights downloaded from the Hugging Face Model Hub."

gr.Interface(
    fn=predict_image,
    inputs=gr.Image(shape=(320, 320), label="Input Image"),
    outputs=gr.Label(num_top_classes=5, label="Top 5 Predictions"),
    title=title,
    description=description,
    article=article,
    theme="huggingface",
    examples=[
        # Examples of classes in CIFAR-100
        # You can add custom images here for better demo experience
        [hf_hub_download(repo_id="gradio/label_image", filename="ferret.jpg", revision="0fe63c3")],
        [hf_hub_download(repo_id="gradio/label_image", filename="cat.jpg", revision="0fe63c3")],
    ]
).launch()
"""
with open("app.py", "w") as f:
    f.write(app_py_content)
```
